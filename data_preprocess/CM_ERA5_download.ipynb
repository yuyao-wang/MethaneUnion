```python
import cdsapi

client = cdsapi.Client()

dataset = 'reanalysis-era5-pressure-levels'
request = {
  'product_type': ['reanalysis'],
  'variable': ['geopotential'],
  'year': ['2024'],
  'month': ['03'],
  'day': ['01'],
  'time': ['13:00'],
  'pressure_level': ['1000'],
  'area': [90, -180, -90, 180],  # North, West, South, East
  'format': 'netcdf',
}
target = 'download.nc'

client.retrieve(dataset, request, target)
```

```python
import xarray as xr

ds = xr.open_dataset("download.nc")  # xarray 会自动用 netcdf4 engine
print(ds)
```

```python
import xarray as xr

ds = xr.open_dataset(
    "download.grib",
    engine="cfgrib",
    backend_kwargs={"filter_by_keys": {"shortName": "z"}}  # 比如你只要 geopotential
)
print(ds)
```
